```python
from roboflow import Roboflow
import os

rf_token = os.environ.get("RF_TOKEN")

rf = Roboflow(api_key=rf_token)
project = rf.workspace("leoni-s-workspace").project("esp32-cars")
version = project.version(1)
dataset = version.download("yolo26")
```

```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")

results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    pretrained=True,
    patience=30,
    lr0=0.001,
    project="esp32-cars",
    name="v1",
)
```

```python
metrics = model.val()
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
```

```python
import sys
import argparse
import cv2
from ultralytics import YOLO


def detect_carts(image_path, model_path="best.pt", output_path="result.jpg", conf=0.80, classes=None):
    """Detects carts in the given image and saves the annotated result."""
    model = YOLO(model_path)

    results = model(image_path, conf=conf, classes=classes)
    result = results[0]

    annotated_image = result.plot()
    cv2.imwrite(output_path, annotated_image)

    count = len(result.boxes)
    print(f"Carts detected: {count}")
    print(f"Image saved to: {output_path}")

    for box in result.boxes:
        class_name = model.names[int(box.cls[0])]
        confidence = float(box.conf[0])
        print(f"  - {class_name} (confidence: {confidence:.2f})")

    return count


detect_carts("img1.png")
```
